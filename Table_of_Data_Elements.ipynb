```python
import pandas as pd
import plotly.express as px

# 1. Load data with the encoding fix for Excel apostrophes
try:
    df = pd.read_csv('List of Data Elements - Sheet1.csv', encoding='cp1252')
except:
    df = pd.read_csv('List of Data Elements - Sheet1.csv', encoding='utf-8')

df.columns = df.columns.str.strip()
df['Element Name'] = df['Element Name'].str.strip()

# 2. THE MANUAL REPLICA MAP (Updated for Business Intel & Cloud names)
manual_coords = {
    "Math": (1,1), "Statistics": (1,2), "Logic": (1,3), "Metadata": (1,4), "Quality": (1,5),
    "Extract": (2,2), "Transform": (2,3), "Pipeline": (2,4), "Orchestra<br>tion": (2,5),
    "SQL": (3,4), "NoSQL": (4,4), "Warehouse": (5,4), "Lake": (6,4), "Cloud": (7,4),
    "Azure": (10,4), "AWS": (11,4), "Google<br>Cloud": (12,4), # Matches your CSV exactly
    "S3": (8,4), "Snowflake": (9,4),
    "Python": (3,5), "R": (4,5), "Jupyter": (5,5), "Git": (6,5), "Pandas": (7,5), "NumPy": (8,5),
    "Imputation": (9,5), "Outliers": (10,5), "Normalize": (11,5), "Dedupe": (12,5),
    "Regression": (13,2), "Classifica<br>tion": (14,2), "Clustering": (15,2), "Neural Nets": (16,2),
    "Rand Forest": (13,3), "XGBoost": (14,3), "SVM": (15,3), "Dec Tree": (16,3),
    "KNN": (13,4), "A/B Test": (14,4), "Hypothesis": (15,4), "P-Value": (16,4), "Conf Int": (13,5),
    "Processing": (14,5), "Dist.<br>Storage": (15,5), "Streaming": (16,5),
    "Value": (18,1), "ROI": (17,2), "Insight": (18,2), "Decision": (17,3), "Story": (18,3),
    "Automate": (17,4), "KPI": (18,4), "Visualizing": (17,5),
    "Business<br>Intel": (18,5), # Fixed name for BI
    "Ethics": (4,8), "Privacy": (5,8), "GDPR": (6,8), "Governance": (7,8), "Security": (8,8)
}

# 3. APPLY COORDINATES
df['x'] = df['Element Name'].map(lambda x: manual_coords.get(x, (None, None))[0])
df['y'] = df['Element Name'].map(lambda x: manual_coords.get(x, (None, None))[1])
df_mapped = df.dropna(subset=['x', 'y']).copy()

# 4. COLOR PALETTE (Updated Governance to a unique Dark Orange)
color_map = {
    "Foundations": "#ff9a9a",
    "Ingestion": "#ffce9a",
    "Storage": "#a0ffa0",
    "Programming": "#9acfff",
    "Cleaning": "#ffffa0",
    "ML": "#c79aff",
    "Advanced Stats": "#a0ffff",
    "Big Data": "#ff9aff",
    "Viz/BI": "#c0c0c0",
    "Outcomes": "#ffcc00",
    "Gov": "#E67E22", # New distinct Dark Orange for Governance
    "Cloud": "#87CEEB"
}

# 5. CREATE THE PLOT
fig = px.scatter(
    df_mapped,
    x="x", y="y", color="Category", text="Symbol",
    color_discrete_map=color_map,
    hover_name="Element Name",
    # 0=Category, 1=Symbol, 2=Definition
    custom_data=["Category", "Symbol", "Definition"]
)

# 6. STYLE TILES
fig.update_traces(
    marker=dict(symbol='square', size=45, line=dict(width=1.5, color='black')),
    # Keeps "Governance" and other long text inside the boxes
    texttemplate="<b>%{text}</b><br><span style='font-size:7px'>%{hovertext}</span>",
    textposition="middle center",

    # The clean, standard hover template
    hovertemplate="""
    <b>%{hovertext}</b><br><br>
    Symbol = %{customdata[1]}<br>
    Category = %{customdata[0]}<br>
    Definition = %{customdata[2]}
    <extra></extra>
    """
)

# 7. LAYOUT
fig.update_layout(
    title=dict(
        text="Mendel's Periodic Table of Data Elements",
        x=0.5, y=0.96, font=dict(size=24)
    ),
    width=1600,
    height=900,
    plot_bgcolor='white',

    # The stable range that doesn't cut off the right side
    xaxis=dict(range=[-2, 20], showgrid=False, zeroline=False, visible=False, fixedrange=True),
    yaxis=dict(range=[10, 0], showgrid=False, zeroline=False, visible=False, fixedrange=True),

    hoverlabel=dict(
        bgcolor="white",
        font_size=13,
        font_family="Arial",
        bordercolor="black",
        align="left"
    ),

    showlegend=True,
    legend=dict(
        title="<b>Category Key</b>",
        x=-0.05,          # The "Goldilocks" position you liked
        y=0.15,
        xanchor="left",
        yanchor="bottom",
        xref="paper",
        bgcolor="rgba(255, 255, 255, 0.7)",
        bordercolor="lightgrey",
        borderwidth=1
    ),

    annotations=[
        dict(
            x=1, y=-0.05, xref="paper", yref="paper",
            text="© 2026 Mendel Bacher | github.com/mendelbacher",
            showarrow=False, font=dict(size=12, color="gray"), xanchor="right"
        )
    ],

    margin=dict(l=120, r=50, t=100, b=80)
)

# --- RUN FUNCTIONS ---
fig.write_html("mendel_periodic_table.html")
fig.show()


```
