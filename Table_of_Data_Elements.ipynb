```python
import pandas as pd
import plotly.express as px

# 1. LOAD DATA
try:
    df = pd.read_csv('List of Data Elements - Sheet1.csv', encoding='cp1252')
except:
    df = pd.read_csv('List of Data Elements - Sheet1.csv', encoding='utf-8')

# Clean whitespace
df.columns = df.columns.str.strip()
df['Element Name'] = df['Element Name'].str.strip()

# 2. YOUR COORDINATES
manual_coords = {
    "Math": (1,1), "Statistics": (1,2), "Logic": (1,3), "Metadata": (1,4), "Quality": (1,5),
    "Extract": (2,2), "Transform": (2,3), "Pipeline": (2,4), "Orchestra<br>tion": (2,5),
    "SQL": (3,4), "NoSQL": (4,4), "Warehouse": (5,4), "Lake": (6,4), "Cloud": (7,4),
    "Azure": (10,4), "AWS": (11,4), "Google<br>Cloud": (12,4), # Matches your CSV exactly
    "S3": (8,4), "Snowflake": (9,4),
    "Python": (3,5), "R": (4,5), "Jupyter": (5,5), "Git": (6,5), "Pandas": (7,5), "NumPy": (8,5),
    "Imputation": (9,5), "Outliers": (10,5), "Normalize": (11,5), "Dedupe": (12,5),
    "Regression": (13,2), "Classify": (14,2), "Clustering": (15,2), "Neural Nets": (16,2),
    "Rand Forest": (13,3), "XGBoost": (14,3), "SVM": (15,3), "Dec Tree": (16,3),
    "KNN": (13,4), "A/B Test": (14,4), "Hypothesis": (15,4), "P-Value": (16,4), "Conf Int": (13,5),
    "Processing": (14,5), "Dist.<br>Storage": (15,5), "Streaming": (16,5),
    "Value": (18,1), "ROI": (17,2), "Insight": (18,2), "Decision": (17,3), "Story": (18,3),
    "Automate": (17,4), "KPI": (18,4), "Visualizing": (17,5),
    "Business<br>Intel": (18,5), # Fixed name for BI
    "Ethics": (4,8), "Privacy": (5,8), "GDPR": (6,8), "Governance": (7,8), "Security": (8,8)
}


# 3. APPLY MAPPING & DISPLAY WRAPPING
df['x'] = df['Element Name'].map(lambda x: manual_coords.get(x, (None, None))[0])
df['y'] = df['Element Name'].map(lambda x: manual_coords.get(x, (None, None))[1])

# This dict handles the line breaks for the visual tiles
display_wrap = {
    "Google Cloud": "Google<br>Cloud",
    "Orchestration": "Orchestra-<br>tion",
    "Business Intel": "Business<br>Intel",
    "Dist. Storage": "Dist.<br>Storage"
}
df['Display Name'] = df['Element Name'].replace(display_wrap)

df_mapped = df.dropna(subset=['x', 'y']).copy()

# 4. CREATE PLOT
df_mapped['Category'] = df_mapped['Category'].str.strip()
fig = px.scatter(
    df_mapped, x="x", y="y", color="Category", text="Symbol",
    color_discrete_map = {
     "Foundations": "#ff9a9a",
    "Ingestion": "#ffce9a",
    "Storage": "#a0ffa0",
    "Programming": "#9acfff",
    "Cleaning": "#ffffa0",
    "ML": "#c79aff",
    "Advanced Stats": "#a0ffff",
    "Big Data": "#ff9aff",
    "Viz/BI": "#c0c0c0",
    "Outcomes": "#ffcc00",
    "Gov": "#D35400",
    "Cloud": "#87CEEB"
},

    hover_name="Element Name",
    custom_data=["Display Name"] # Corrected property name
)

# 5. STYLE TILES
fig.update_traces(
    marker=dict(symbol='square', size=45, line=dict(width=1.5, color='black')),
    # customdata[0] refers to the "Display Name" we passed in step 4
    texttemplate="<b>%{text}</b><br><span style='font-size:7px'>%{customdata[0]}</span>",
    textposition="middle center"
)

# 6. LAYOUT
fig.update_layout(
    title=dict(text="Mendel's Periodic Table of Data Elements", x=0.5, y=0.96, font=dict(size=24)),
    width=1650, height=900, plot_bgcolor='white',
    xaxis=dict(range=[-1, 21], visible=False),
    yaxis=dict(range=[10, 0], visible=False),
    legend=dict(title="<b>Category Key</b>", x=-0.07, y=0.15, xanchor="left", yanchor="bottom", borderwidth=1),
    margin=dict(l=120, r=50, t=100, b=100),
    annotations=[dict(x=1, y=-0.08, xref="paper", yref="paper", text="© 2026 Mendel Bacher", showarrow=False, font=dict(size=14, color="gray"), xanchor="right")]
)

fig.show()
```
